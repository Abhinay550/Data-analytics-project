```python
# Case Study 02 - Seaborn Techniques on MIDMARKS

# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
data_file = "MIDMARKS.xlsx"
df = pd.read_excel(data_file)

# Display basic information about the dataset
print("Dataset Info:")
df.info()
print("\nFirst few rows of the dataset:")
print(df.head())

# Fill missing numeric values with the mean and drop rows with missing categorical data
df.fillna(df.mean(numeric_only=True), inplace=True)
df.dropna(subset=['SECTION'], inplace=True)

# Convert numeric columns to appropriate types
numeric_columns = ["DV", "M-II", "PP", "BEEE", "FL", "FIMS"]
for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# 1. Adjusted Seaborn Theme
sns.set(style="whitegrid", palette="pastel")



```

```python
# 2. Countplot for Students by Section
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="SECTION")
plt.title("Number of Students in Each Section")
plt.xlabel("Section")
plt.ylabel("Count")
plt.show()
```

```python
# 3. Distribution of FL Scores
plt.figure(figsize=(8, 5))
sns.histplot(df['FL'], kde=True, bins=12, color="green", edgecolor="black")
plt.title("Distribution of FL Scores")
plt.xlabel("FL Scores")
plt.ylabel("Frequency")
plt.show()
```

```python
# 4. Average BEEE Scores by Section
avg_scores = df.groupby("SECTION")["BEEE"].mean().reset_index()
plt.figure(figsize=(8, 5))
sns.barplot(data=avg_scores, x="SECTION", y="BEEE", ci=None, palette="coolwarm")
plt.title("Average BEEE Scores by Section")
plt.xlabel("Section")
plt.ylabel("Average BEEE Scores")
plt.show()
```

```python
# 5. Pairplot of Numeric Data
pairplot_cols = numeric_columns + ["SECTION"]
plt.figure(figsize=(10, 6))
sns.pairplot(df[pairplot_cols], hue="SECTION", diag_kind="kde")
plt.suptitle("Pairplot of Numeric Columns by Section", y=1.02)
plt.show()
```

```python
# 6. Violin Plot: Distribution of PP Scores by Section
plt.figure(figsize=(8, 5))
sns.violinplot(data=df, x="SECTION", y="PP", inner="quartile", palette="muted")
plt.title("Violin Plot of PP Scores by Section")
plt.xlabel("Section")
plt.ylabel("PP Scores")
plt.show()
```

```python
# 9. Scatterplot: M-II vs FIMS
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="M-II", y="FIMS", hue="SECTION", size="FL", sizes=(20, 200), alpha=0.7)
plt.title("Scatterplot: M-II vs FIMS by Section")
plt.xlabel("M-II Scores")
plt.ylabel("FIMS Scores")
plt.legend(title="Section")
plt.show()
```

```python
# 10. KDE Plot for FL Scores by Section
plt.figure(figsize=(8, 5))
sns.kdeplot(data=df, x="FL", hue="SECTION", fill=True, common_norm=False, alpha=0.4, linewidth=1.5)
plt.title("KDE Plot of FL Scores by Section")
plt.xlabel("FL Scores")
plt.ylabel("Density")
plt.show()
```

```python
# 11. Residual Plot for BEEE and FIMS
plt.figure(figsize=(8, 5))
sns.residplot(data=df, x="BEEE", y="FIMS", lowess=True, color="purple")
plt.title("Residual Plot: BEEE vs FIMS")
plt.xlabel("BEEE Scores")
plt.ylabel("Residuals")
plt.show()
```

```python
# 12. Boxen Plot for All Scores by Section
plt.figure(figsize=(10, 6))
sns.boxenplot(data=df.melt(id_vars="SECTION", value_vars=numeric_columns), x="SECTION", y="value", hue="variable", palette="Set2")
plt.title("Boxen Plot of Scores by Section")
plt.xlabel("Section")
plt.ylabel("Scores")
plt.show()
```

```python
# 13. Enhanced Bar Plot for Section Averages
section_means = df.groupby("SECTION").mean().reset_index()
section_means_melted = section_means.melt(id_vars="SECTION", value_vars=numeric_columns)
plt.figure(figsize=(12, 6))
sns.barplot(data=section_means_melted, x="variable", y="value", hue="SECTION", palette="Paired")
plt.title("Section-Wise Average Scores Across Subjects")
plt.xlabel("Subjects")
plt.ylabel("Average Scores")
plt.legend(title="Section")
plt.show()

```

```python
# 14. Catplot for FIMS by Section
sns.catplot(data=df, x="SECTION", y="FIMS", kind="swarm", aspect=2, height=6, palette="Spectral")
plt.title("FIMS Scores Distribution by Section")
plt.show()

```

```python
# 15. Stacked Bar Chart for Total Scores by Section
total_scores = df.groupby("SECTION")[numeric_columns].sum().reset_index()
total_scores.set_index("SECTION").plot(kind="bar", stacked=True, colormap="viridis", figsize=(10, 6))
plt.title("Total Scores by Section")
plt.xlabel("Section")
plt.ylabel("Total Scores")
plt.show()
```
